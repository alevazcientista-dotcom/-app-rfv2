# **Módulo 3 | SQL: Selecionando & Ordenando**
## Notebook corrigido para rodar no Google Colab

Este notebook não usa AWS nem upload de arquivo.  
Ele recria a tabela `transacoes` dentro do próprio Colab usando SQLite e executa as queries pedidas pela atividade.


## 1. Preparação dos dados

```python

import pandas as pd
import sqlite3

# Dados do exercício M3 - transacoes
transacoes = pd.DataFrame({
    "id_cliente":   [1, 2, 3, 1, 1, 3, 4, 5, 3],
    "id_transacao": [768805383, 768805399, 818770008, 76856563, 767573759, 818575758, 764545534, 76766789, 8154567758],
    "valor":        [50.74, 30.90, 110.00, 2000.90, 15.70, 2.99, 50.74, 10.00, 1100.00],
    "id_loja":      ["magalu", "giraffas", "postoshell", "magalu", "subway", "seveneleven", "extra", "subway", "shopee"]
})

# Cria banco SQLite em memória para simular as consultas SQL
con = sqlite3.connect(":memory:")
transacoes.to_sql("transacoes", con, index=False, if_exists="replace")

print("✅ Tabela 'transacoes' criada com sucesso!")
print(f"{len(transacoes)} linhas carregadas.")
display(transacoes)
```

## 2. Query 1 — visualizar todos os dados

```python

# Query 1
# SELECT * FROM transacoes

query1 = pd.read_sql_query("""
SELECT *
FROM transacoes
""", con)

display(query1)
query1.to_csv("query1.csv", index=False)
print("✅ Arquivo query1.csv salvo")
```

## 3. Query 2 — usar alias `AS`

```python

# Query 2
# SELECT id_cliente, valor, id_loja AS nome_loja FROM transacoes;

query2 = pd.read_sql_query("""
SELECT id_cliente, valor, id_loja AS nome_loja
FROM transacoes
""", con)

display(query2)
query2.to_csv("query2.csv", index=False)
print("✅ Arquivo query2.csv salvo")
```

## 4. Query 3 — lojas sem repetição com `DISTINCT`

```python

# Query 3
# SELECT DISTINCT id_loja AS nome_loja FROM transacoes;

query3 = pd.read_sql_query("""
SELECT DISTINCT id_loja AS nome_loja
FROM transacoes
""", con)

display(query3)
query3.to_csv("query3.csv", index=False)
print("✅ Arquivo query3.csv salvo")
```

## 5. Query 4 — ordenar e limitar resultados

```python

# Query 4
# SELECT id_cliente, valor FROM transacoes ORDER BY valor DESC LIMIT 2;

query4 = pd.read_sql_query("""
SELECT id_cliente, valor
FROM transacoes
ORDER BY valor DESC
LIMIT 2
""", con)

display(query4)
query4.to_csv("query4.csv", index=False)
print("✅ Arquivo query4.csv salvo")
```

## 6. Conferência final

```python

# Conferência final
import os

arquivos = ["query1.csv", "query2.csv", "query3.csv", "query4.csv"]

print("Arquivos gerados:")
for arquivo in arquivos:
    print(f" - {arquivo}: {'OK' if os.path.exists(arquivo) else 'ERRO'}")

print("\n🎉 M3 concluído! Agora compartilhe o link deste Colab na EBAC.")
```
