```python
# M4 — SQL: Filtrando & Seleção Condicional
# Exercício completo — SEM UPLOAD DE ARQUIVO
# Abra no Colab e execute tudo (Ctrl+F9)


import pandas as pd
import sqlite3
import numpy as np


# DADOS — transacoes (gerados automaticamente, sem upload)

np.random.seed(42)
n = 500

lojas = ['magalu','extra','giraffas','subway','postoshell','seveneleven','hering','amazon']
pesos = [0.20, 0.15, 0.12, 0.13, 0.10, 0.08, 0.12, 0.10]

transacoes = pd.DataFrame({
    'id_cliente':   np.random.randint(1000, 9999, n),
    'id_transacao': np.arange(10001, 10001 + n),
    'valor':        np.round(np.random.uniform(5.0, 1500.0, n), 2),
    'id_loja':      np.random.choice(lojas, n, p=pesos)
})

# Carrega no SQLite (simula o Athena)
con = sqlite3.connect(':memory:')
transacoes.to_sql('transacoes', con, index=False, if_exists='replace')

print(" Tabela 'transacoes' criada com sucesso!")
print(f"   {len(transacoes)} linhas | colunas: {list(transacoes.columns)}")
print(transacoes.head(3))

```

```python
# QUERY 1 — AND

print("\n" + "="*55)
print("QUERY 1 — valor > 30 AND id_loja = 'magalu'")
print("="*55)

query1 = pd.read_sql_query("""
    SELECT *
    FROM transacoes
    WHERE valor > 30 AND id_loja = 'magalu'
""", con)

print(query1)
print(f"\nTotal: {len(query1)} linhas")
query1.to_csv('query1.csv', index=False)
print("query1.csv salvo")

# ============================================================
```

```python
# QUERY 2 — OR

print("\n" + "="*55)
print("QUERY 2 — valor > 30 OR id_loja = 'magalu'")
print("="*55)

query2 = pd.read_sql_query("""
    SELECT *
    FROM transacoes
    WHERE valor > 30 OR id_loja = 'magalu'
""", con)

print(query2)
print(f"\nTotal: {len(query2)} linhas")
query2.to_csv('query2.csv', index=False)
print("query2.csv salvo")

```

```python
# QUERY 3 — IN

print("\n" + "="*55)
print("QUERY 3 — id_loja IN ('magalu','subway') AND valor > 10")
print("="*55)

query3 = pd.read_sql_query("""
    SELECT *
    FROM transacoes
    WHERE id_loja IN ('magalu','subway') AND valor > 10
""", con)

print(query3)
print(f"\nTotal: {len(query3)} linhas")
query3.to_csv('query3.csv', index=False)
print(" query3.csv salvo")


```

```python
# QUERY 4 — BETWEEN

print("\n" + "="*55)
print("QUERY 4 — valor BETWEEN 60.0 AND 1000.0")
print("="*55)

query4 = pd.read_sql_query("""
    SELECT *
    FROM transacoes
    WHERE valor BETWEEN 60.0 AND 1000.0
""", con)

print(query4)
print(f"\nTotal: {len(query4)} linhas")
query4.to_csv('query4.csv', index=False)
print("query4.csv salvo")

```

```python
# QUERY 5 — LIKE com %

print("\n" + "="*55)
print("QUERY 5 — id_loja LIKE 'mag%'")
print("="*55)

query5 = pd.read_sql_query("""
    SELECT *
    FROM transacoes
    WHERE id_loja LIKE 'mag%'
""", con)

print(query5)
print(f"\nTotal: {len(query5)} linhas")
query5.to_csv('query5.csv', index=False)
print("query5.csv salvo")
```

```python
# QUERY 6 — LIKE com %x%

print("\n" + "="*55)
print("QUERY 6 — id_loja LIKE '%sh%'")
print("="*55)

query6 = pd.read_sql_query("""
    SELECT *
    FROM transacoes
    WHERE id_loja LIKE '%sh%'
""", con)

print(query6)
print(f"\nTotal: {len(query6)} linhas")
query6.to_csv('query6.csv', index=False)
print("query6.csv salvo")
```

```python
# QUERY 7 — CASE / Seleção Condicional

print("\n" + "="*55)
print("QUERY 7 — CASE: classeValor e tipo_compra")
print("="*55)

query7 = pd.read_sql_query("""
    SELECT id_cliente, id_loja, valor,
    CASE
        WHEN valor > 1000 THEN 'Compra com alto valor'
        WHEN valor < 1000 THEN 'Compra com baixo valor'
    END AS classeValor,
    CASE
        WHEN id_loja IN ('giraffas','subway')       THEN 'alimentacao'
        WHEN id_loja IN ('magalu','extra')           THEN 'variedade'
        WHEN id_loja IN ('postoshell','seveneleven') THEN '24horas'
        ELSE 'outros'
    END AS tipo_compra
    FROM transacoes
""", con)

print(query7)
print(f"\nTotal: {len(query7)} linhas")
query7.to_csv('query7.csv', index=False)
print("query7.csv salvo")
```

```python
# RESUMO FINAL

print("\n" + "="*55)
print("RESUMO — todas as queries executadas")
print("="*55)
for i, q in enumerate([query1,query2,query3,query4,query5,query6,query7], 1):
    print(f"  query{i}.csv → {len(q):>4} linhas")
```
