````python
# Importa a biblioteca pandasql
from pandasql import sqldf
````

````python
# IMPORTAÇÃO DAS BIBLIOTECAS

import pandas as pd
import numpy as np

# Configurações de visualização
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)


# FUNÇÕES PARA CARREGAMENTO E ANÁLISE DE DADOS

def load_and_display_data(file_path):
    """Carrega o CSV em um DataFrame e exibe informações iniciais."""
    print(f"Carregando dados de: {file_path}")
    # Explicitly set header=None as the first row of data seems to be incorrectly used as header
    df = pd.read_csv(file_path, header=None)

    # Assign column names based on the assumed structure from the context
    column_names = [
        'id', 'idade', 'sexo', 'dependentes', 'escolaridade',
        'estado_civil', 'limite_credito', 'valor_transacao_total_12m',
        'qtd_transacoes_12m'
    ]
    df.columns = column_names

    print("\nPrimeiras linhas:")
    display(df.head())

    print("\nInformações gerais:")
    df.info()

    print("\nEstatísticas descritivas:")
    display(df.describe())

    return df

def display_all_records(df):
    """Consulta 1: Visualiza todos os registros.
    Equivalente ao: SELECT * FROM clientes"""
    print("\nCONSULTA 1: Todos os registros")
    display(df)

def get_male_clients_by_age(df):
    """Consulta 2: Clientes masculinos ordenados por idade.
    Equivalente ao:
    SELECT id, idade, limite_credito
    FROM clientes
    WHERE sexo='M'
    ORDER BY idade DESC
    """
    clientes_masculinos = (
        df[df['sexo'] == 'M']
        [['id', 'idade', 'limite_credito']]
        .sort_values(by='idade', ascending=False)
    )
    print("\nCONSULTA 2: Clientes masculinos ordenados por idade:")
    display(clientes_masculinos)
    return clientes_masculinos

def get_average_age_by_sex(df):
    """Consulta 3: Média de idade por sexo.
    Equivalente ao:
    SELECT sexo, AVG(idade)
    FROM clientes
    GROUP BY sexo
    """
    media_idade = (
        df.groupby('sexo')['idade']
        .mean()
        .round(0)
        .astype(int)
        .reset_index()
    )
    media_idade.columns = ['sexo', 'media_idade']
    print("\nCONSULTA 3: Média de idade por sexo:")
    display(media_idade)
    return media_idade

def analyze_extra_metrics(df):
    """Realiza análises extras sobre o DataFrame."""
    print("\nANÁLISES EXTRAS")

    print("\nQuantidade de clientes por sexo:")
    display(df['sexo'].value_counts())

    print("\nLimite de crédito médio por sexo:")
    display(
        df.groupby('sexo')['limite_credito']
        .mean()
        .round(2)
    )

    print("\nQuantidade média de transações por sexo:")
    display(
        df.groupby('sexo')['qtd_transacoes_12m']
        .mean()
        .round(2)
    )

    print("\nTop 10 maiores limites de crédito:")
    display(
        df[['id','sexo','idade','limite_credito']]
        .sort_values('limite_credito', ascending=False)
        .head(10)
    )


# FLUXO PRINCIPAL DE EXECUÇÃO

# Substitua pelo nome do arquivo caso necessário
file_name = 'SQL para análise de dados M1 Exercício credito-aula.csv'

df_credito = load_and_display_data(file_name)

if df_credito is not None:
    display_all_records(df_credito)
    get_male_clients_by_age(df_credito)
    get_average_age_by_sex(df_credito)
    analyze_extra_metrics(df_credito)

print("\nAnálise concluída com sucesso!")
````

Para realizar consultas SQL, definimos uma função auxiliar `pysqldf` que executa a consulta no contexto do DataFrame `df_credito`.

````python
pysqldf = lambda q: sqldf(q, globals())
````

### CONSULTA 1 (SQL): Visualizar todos os registros
Equivalente ao: `SELECT * FROM df_credito`

````python
query_all_records = """
SELECT *
FROM df_credito
"""

print("\nCONSULTA 1 (SQL): Todos os registros")
display(pysqldf(query_all_records))
````

### CONSULTA 2 (SQL): Clientes masculinos ordenados por idade
Equivalente ao:

```sql
SELECT id, idade, limite_credito
FROM df_credito
WHERE sexo='M'
ORDER BY idade DESC
```

````python
query_male_clients = """
SELECT id, idade, limite_credito
FROM df_credito
WHERE sexo = 'M'
ORDER BY idade DESC
"""

print("\nCONSULTA 2 (SQL): Clientes masculinos ordenados por idade:")
display(pysqldf(query_male_clients))
````

### CONSULTA 3 (SQL): Média de idade por sexo
Equivalente ao:

```sql
SELECT sexo,
       AVG(idade) as media_idade
FROM df_credito
GROUP BY sexo
```

````python
query_avg_age_by_sex = """
SELECT sexo,
       CAST(ROUND(AVG(idade), 0) AS INT) AS media_idade
FROM df_credito
GROUP BY sexo
"""

print("\nCONSULTA 3 (SQL): Média de idade por sexo:")
display(pysqldf(query_avg_age_by_sex))
````

### ANÁLISES EXTRAS (SQL)

````python
query_clients_by_sex = """
SELECT sexo, COUNT(id) as quantidade_clientes
FROM df_credito
GROUP BY sexo
"""

print("\nQuantidade de clientes por sexo:")
display(pysqldf(query_clients_by_sex))
````

````python
query_avg_credit_limit = """
SELECT sexo, ROUND(AVG(limite_credito), 2) as limite_credito_medio
FROM df_credito
GROUP BY sexo
"""

print("\nLimite de crédito médio por sexo:")
display(pysqldf(query_avg_credit_limit))
````

````python
query_avg_transactions = """
SELECT sexo, ROUND(AVG(qtd_transacoes_12m), 2) as media_transacoes
FROM df_credito
GROUP BY sexo
"""

print("\nQuantidade média de transações por sexo:")
display(pysqldf(query_avg_transactions))
````

````python
query_top_10_credit = """
SELECT id, sexo, idade, limite_credito
FROM df_credito
ORDER BY limite_credito DESC
LIMIT 10
"""

print("\nTop 10 maiores limites de crédito:")
display(pysqldf(query_top_10_credit))
````
